```python
# Separate features and target
X = df.drop("Outcome", axis=1)
y = df["Outcome"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
# Columns where zero can be considered an invalid/missing measurement
zero_columns = [
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI"
]

# Replace zero values with NaN
X[zero_columns] = X[zero_columns].replace(0, np.nan)

# Check missing values after replacement
print(X.isnull().sum())
```

```python
from sklearn.model_selection import train_test_split, StratifiedKFold, GridSearchCV
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
```

```python
# Split dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
from lightgbm import LGBMClassifier
from xgboost import XGBClassifier
```

```python
# Create LightGBM pipeline
lgbm_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="median")
    ),
    (
        "model",
        LGBMClassifier(
            random_state=42,
            verbosity=-1
        )
    )
])

# Train LightGBM model
lgbm_pipeline.fit(X_train, y_train)

# Make predictions
y_pred_lgbm = lgbm_pipeline.predict(X_test)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)
```

```python
# Calculate LightGBM evaluation metrics
lgbm_accuracy = accuracy_score(y_test, y_pred_lgbm)
lgbm_precision = precision_score(y_test, y_pred_lgbm)
lgbm_recall = recall_score(y_test, y_pred_lgbm)
lgbm_f1 = f1_score(y_test, y_pred_lgbm)

print("LightGBM Performance")
print("--------------------")
print("Accuracy :", lgbm_accuracy)
print("Precision:", lgbm_precision)
print("Recall   :", lgbm_recall)
print("F1 Score :", lgbm_f1)
```

```python
print("LightGBM Classification Report")
print(
    classification_report(
        y_test,
        y_pred_lgbm
    )
)
```

```python
# Calculate confusion matrix
cm_lgbm = confusion_matrix(y_test, y_pred_lgbm)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm_lgbm,
    annot=True,
    fmt="d"
)

plt.title("LightGBM Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()
```

```python
# Create XGBoost pipeline
xgb_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="median")
    ),
    (
        "model",
        XGBClassifier(
            random_state=42,
            eval_metric="logloss"
        )
    )
])

# Train XGBoost model
xgb_pipeline.fit(X_train, y_train)

# Make predictions
y_pred_xgb = xgb_pipeline.predict(X_test)
```

```python
# Calculate XGBoost evaluation metrics
xgb_accuracy = accuracy_score(y_test, y_pred_xgb)
xgb_precision = precision_score(y_test, y_pred_xgb)
xgb_recall = recall_score(y_test, y_pred_xgb)
xgb_f1 = f1_score(y_test, y_pred_xgb)

print("XGBoost Performance")
print("-------------------")
print("Accuracy :", xgb_accuracy)
print("Precision:", xgb_precision)
print("Recall   :", xgb_recall)
print("F1 Score :", xgb_f1)
```

```python
print("XGBoost Classification Report")
print(
    classification_report(
        y_test,
        y_pred_xgb
    )
)
```

```python
# Calculate confusion matrix
cm_xgb = confusion_matrix(y_test, y_pred_xgb)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm_xgb,
    annot=True,
    fmt="d"
)

plt.title("XGBoost Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()
```

```python
# Create a comparison DataFrame
results = pd.DataFrame({
    "Model": ["LightGBM", "XGBoost"],
    "Accuracy": [lgbm_accuracy, xgb_accuracy],
    "Precision": [lgbm_precision, xgb_precision],
    "Recall": [lgbm_recall, xgb_recall],
    "F1 Score": [lgbm_f1, xgb_f1]
})

results
```

```python
# Convert results into a format suitable for plotting
results_plot = results.set_index("Model")

# Plot comparison
results_plot.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Performance Comparison: LightGBM vs XGBoost")
plt.xlabel("Model")
plt.ylabel("Score")
plt.ylim(0, 1)

plt.xticks(rotation=0)
plt.legend(loc="lower right")

plt.show()
```

```python
# Define Stratified K-Fold cross-validation
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

```python
from sklearn.model_selection import cross_val_score

# Perform 5-fold cross-validation for LightGBM
lgbm_cv_scores = cross_val_score(
    lgbm_pipeline,
    X,
    y,
    cv=cv,
    scoring="f1"
)

print("LightGBM Cross-Validation F1 Scores:")
print(lgbm_cv_scores)

print("Mean LightGBM F1 Score:",
      lgbm_cv_scores.mean())
```

```python
# Perform 5-fold cross-validation for XGBoost
xgb_cv_scores = cross_val_score(
    xgb_pipeline,
    X,
    y,
    cv=cv,
    scoring="f1"
)

print("XGBoost Cross-Validation F1 Scores:")
print(xgb_cv_scores)

print("Mean XGBoost F1 Score:",
      xgb_cv_scores.mean())
```

```python
# Define LightGBM parameter grid
lgbm_params = {
    "model__n_estimators": [100, 200],
    "model__learning_rate": [0.05, 0.1],
    "model__num_leaves": [15, 31],
    "model__max_depth": [-1, 5]
}

# GridSearchCV
lgbm_grid = GridSearchCV(
    estimator=lgbm_pipeline,
    param_grid=lgbm_params,
    cv=5,
    scoring="f1",
    n_jobs=-1
)

# Train using grid search
lgbm_grid.fit(X_train, y_train)

print("Best LightGBM Parameters:")
print(lgbm_grid.best_params_)

print("Best LightGBM CV F1 Score:")
print(lgbm_grid.best_score_)
```

```python
# Define XGBoost parameter grid
xgb_params = {
    "model__n_estimators": [100, 200],
    "model__learning_rate": [0.05, 0.1],
    "model__max_depth": [3, 5],
    "model__subsample": [0.8, 1.0]
}

# GridSearchCV
xgb_grid = GridSearchCV(
    estimator=xgb_pipeline,
    param_grid=xgb_params,
    cv=5,
    scoring="f1",
    n_jobs=-1
)

# Train using grid search
xgb_grid.fit(X_train, y_train)

print("Best XGBoost Parameters:")
print(xgb_grid.best_params_)

print("Best XGBoost CV F1 Score:")
print(xgb_grid.best_score_)
```

```python
# Predictions from tuned models
y_pred_lgbm_tuned = lgbm_grid.predict(X_test)
y_pred_xgb_tuned = xgb_grid.predict(X_test)

# LightGBM tuned metrics
lgbm_tuned_results = [
    accuracy_score(y_test, y_pred_lgbm_tuned),
    precision_score(y_test, y_pred_lgbm_tuned),
    recall_score(y_test, y_pred_lgbm_tuned),
    f1_score(y_test, y_pred_lgbm_tuned)
]

# XGBoost tuned metrics
xgb_tuned_results = [
    accuracy_score(y_test, y_pred_xgb_tuned),
    precision_score(y_test, y_pred_xgb_tuned),
    recall_score(y_test, y_pred_xgb_tuned),
    f1_score(y_test, y_pred_xgb_tuned)
]

# Create final comparison table
final_results = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score"
    ],
    "LightGBM": lgbm_tuned_results,
    "XGBoost": xgb_tuned_results
})

final_results
```

```python
# Plot final model comparison
final_results.set_index("Metric").plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Final Performance: Tuned LightGBM vs Tuned XGBoost")
plt.xlabel("Evaluation Metric")
plt.ylabel("Score")
plt.ylim(0, 1)

plt.xticks(rotation=0)

plt.show()
```

```python
# Calculate average score across all evaluation metrics
lgbm_average = np.mean(lgbm_tuned_results)
xgb_average = np.mean(xgb_tuned_results)

print("Average LightGBM Score:", lgbm_average)
print("Average XGBoost Score:", xgb_average)

if lgbm_average > xgb_average:
    print("\nLightGBM performed better overall.")
elif xgb_average > lgbm_average:
    print("\nXGBoost performed better overall.")
else:
    print("\nBoth models performed equally overall.")
```

```python
# Get the trained LightGBM model from the tuned pipeline
lgbm_model = lgbm_grid.best_estimator_.named_steps["model"]

# Get feature importance
lgbm_importance = pd.Series(
    lgbm_model.feature_importances_,
    index=X.columns
).sort_values(ascending=False)

# Plot
plt.figure(figsize=(10, 6))

lgbm_importance.plot(kind="bar")

plt.title("LightGBM Feature Importance")
plt.xlabel("Features")
plt.ylabel("Importance")

plt.xticks(rotation=45)

plt.show()
```

```python
# Get the trained XGBoost model
xgb_model = xgb_grid.best_estimator_.named_steps["model"]

# Get feature importance
xgb_importance = pd.Series(
    xgb_model.feature_importances_,
    index=X.columns
).sort_values(ascending=False)

# Plot
plt.figure(figsize=(10, 6))

xgb_importance.plot(kind="bar")

plt.title("XGBoost Feature Importance")
plt.xlabel("Features")
plt.ylabel("Importance")

plt.xticks(rotation=45)

plt.show()
```
